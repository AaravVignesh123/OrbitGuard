# Session 1 — ISS position, right now

**Phase 0 · tooling fluency.** Load a real satellite from a live TLE, propagate it to the current instant, and print its sub-point (the lat/lon on Earth directly beneath it) and altitude.

*Concept:* a **TLE** (two-line element set) encodes an object's orbit at a reference *epoch*. skyfield feeds it to the **SGP4** propagator, which advances the orbit to any time. `subpoint()` converts the inertial position to geodetic lat/lon/alt.

```python
from skyfield.api import load

ts = load.timescale()
url = "https://celestrak.org/NORAD/elements/gp.php?GROUP=stations&FORMAT=tle"
sats = load.tle_file(url)
iss = next(s for s in sats if "ISS" in s.name)
print(iss)
```

```python
t = ts.now()
sp = iss.at(t).subpoint()
print(f"ISS is over lat {sp.latitude.degrees:+.2f}\u00b0, "
      f"lon {sp.longitude.degrees:+.2f}\u00b0, "
      f"alt {sp.elevation.km:.1f} km")
# Sanity check: altitude should be ~400 km. Re-run in a minute; lat/lon move.
```

**Done when:** the printed altitude is ~400 km and the lat/lon change on re-run. This is the atom the whole catalog is built from.
